## Reading the result

- The edge phase rotations cancel exactly around the triangle.
- For the pure generalized-W state, all three population responses are negative, so their odd product flips the Z response.
- The loop singular values are all one, proving the observed loop defect is not attenuation.
- With white noise, the orientation class can change only when an edge population response crosses zero, producing rank loss at the transition.
